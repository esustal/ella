```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

exoplanet_df = pd.read_csv("exoplanet_data.csv", comment="#")
exoplanet_df
```

```python
exoplanet_df.shape
```

```python
print(exoplanet_df.columns)

# sy_kmag	sy_kmagerr1	sy_kmagerr2	sy_gaiamag	sy_gaiamagerr1	sy_gaiamagerr2
```

```python
# Table Label: Planet Name
# Database Column Name: pl_name
# Description: Planet name most commonly used in the literature.

# Table Label: Host Name
# Database Column Name: hostname
# Description: Name of the star that the planet orbits.

# Table Label: Discovery Method
# Database Column Name: discoverymethod
# Description: Method by which the planet was first identified.

# Table Label: Spectral Type
# Database Column Name: st_spectype
# Description: Classification of the host star based on its spectral characteristics.

# Table Label: Equilibrium Temperature [K]
# Database Column Name: pl_eqt
# Description: Estimated temperature of the planet assuming it behaves like a blackbody heated by its host star.

# Table Label: Planet Mass or Mass*sin(i) [Jupiter Mass]
# Database Column Name: pl_bmassj
# Description: Best available estimate of the planet's mass, or minimum mass, measured in Jupiter masses.
```

```python
exoplanet_df = exoplanet_df.dropna()
exoplanet_df
```

```python
exoplanet_df.shape
```

```python
exoplanet_df.head()
```

```python
exoplanet_df.tail()
```

```python
exoplanet_df.columns
```

```python
exoplanet_df["discoverymethod"].value_counts()
print(exoplanet_df["discoverymethod"].value_counts())
```

```python
counts = exoplanet_df["discoverymethod"].value_counts()

print("Most:", counts.idxmax())
print("Least:", counts.idxmin())
```

```python
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data = exoplanet_df,
    x = "pl_bmassj",
    y = "pl_eqt",
    hue = "discoverymethod"
)

plt.title("Planet Mass vs. Equilibrium Temperature")
plt.xlabel("Planet Mass (Jupiter Masses)")
plt.ylabel("Equilibrium Temperature (K)")
plt.legend(title="Discovery Method")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=exoplanet_df,
    x=np.log(exoplanet_df["pl_bmassj"]),
    y=np.log(exoplanet_df["pl_eqt"]),
    hue="discoverymethod"
)

plt.title("Log Planet Mass vs. Log Equilibrium Temperature")
plt.xlabel("Log Planet Mass")
plt.ylabel("Log Equilibrium Temperature")
plt.legend(title="Discovery Method")

plt.show()
```
