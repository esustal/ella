```python
import numpy as np

arr = np.array ([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def prime_rows(arr):
    rows = []
    for row in arr:
            for n in row:
                if n in [2, 3, 5, 7, 11, 13, 17, 19]:
                    rows.append(row)
                    break
    return np.array(rows)

print(prime_rows(arr))
```

```python
import numpy as np

def checkerboard():
    board = np.zeros((8,8), dtype = int)

    for i in range(0, 8, 2):
        board[i, 1::2] = 1

    return board

print(checkerboard())


```

```python
import numpy as np

def expansion(word, n):
    return (" " * n).join(word)

print(expansion("galaxy clusters", 0))
print(expansion("galaxy clusters", 1))
print(expansion("galaxy clusters", 2))
```

```python
import numpy as np

stars = np.array([[1123, 1456, 1789, 1324, 1876],
[1567, 1987, 1678, 1405, 1589],
[1345, 1654, 1523, 1109, 1923],
[1298, 1890, 1367, 1784, 1432],
[1823, 1756, 1489, 1672, 1550]])

def  second_dimmest(stars):
    sorted_stars = np.sort(stars, axis = 1)
    return sorted_stars[:, -2]

print(second_dimmest(stars))
```
